```python
import torch
```

```python
zeros_tensor  = torch.zeros(5)
print(zeros_tensor)
```

```python
zeros_tensor = torch.zeros(256,256)
print(zeros_tensor)
print(zeros_tensor.shape)
```

```python
import matplotlib.pyplot as plt
plt.imshow(zeros_tensor.numpy(),cmap='grey',vmin=0,vmax=1)
plt.title("Zeros Tensor Dark Image")
plt.axis("off")
plt.show()
```

```python
ones_tensor = torch.ones(5)
print("Ones Tensor:",ones_tensor)
print("Dimensions",ones_tensor.ndim)
print("Shape:",ones_tensor.shape)
ones_tensor = torch.ones(256,256,3)
plt.imshow(ones_tensor.numpy(),vmin=0,vmax=1)
plt.title("Ones Tensor White Image")
plt.axis("off")
```
