```python

from ultralytics import YOLO

# Load your model
model = YOLO("yolov8n.pt")

# Map Class ID 32 ("sports ball") to "soccer ball"
model.names[32] = "soccer ball"

# Run inference
results = model("/content/drive/MyDrive/bicycle.jpeg")

# Visualize results - the box will now say "soccer ball"
for r in results:
    for box in r.boxes:
        if int(box.cls[0])==32:
            r.names[32]="soccer ball"
    r.show()
```
