```python
import torch
```

```python
tensor = torch.arange(12)
```

```python
reshaped_tensor_with_view=tensor.view(2,6)
reshaped_tensor_with_reshape=tensor.reshape(2,6)
print(reshaped_tensor_with_view)
print(reshaped_tensor_with_view.shape)
print(reshaped_tensor_with_reshape)
```

```python
print("tensor is contiguous or not:",tensor.is_contiguous())
```

```python
inferred_shape=tensor.view(3,-1)
print(inferred_shape)
print(inferred_shape.size())
```

```python
tensor_3d=torch.arange(24).reshape(2,3,4)
print(tensor_3d)
```

```python
flattened=tensor_3d.view(-1)
print(flattened)
print("is_contiguous:",flattened.is_contiguous())
```

```python
tensor_2d=torch.arange(24).reshape(12,2)
transposed_tensor=tensor_2d.t()
print(transposed_tensor.is_contiguous())
```
