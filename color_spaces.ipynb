```python
import cv2
```

```python
bgr_image = cv2.imread("./istockphoto-958518084-612x612.jpg")
```

```python
#Convert to greyscale

greyImage=cv2.cvtColor(bgr_image,cv2.COLOR_BGR2GRAY)
```

```python
cv2.imwrite("./greyImage.jpg",greyImage)
```

```python
# Convert To HSV
hsvImage=cv2.cvtColor(bgr_image,cv2.COLOR_BGR2HSV) 
cv2.imwrite("./hsvImage.jpg",hsvImage)
```

```python
# Convert To LAB
labImage=cv2.cvtColor(bgr_image,cv2.COLOR_BGR2LAB) 
cv2.imwrite("./labImage.jpg",labImage)
```

```python
# Convert To RGB
rgbImage=cv2.cvtColor(bgr_image,cv2.COLOR_BGR2RGB) 
cv2.imwrite("./rgbImage.jpg",rgbImage)
```
