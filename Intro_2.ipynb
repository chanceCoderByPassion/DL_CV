```python
# Accessing elements in a tensor
'''
tensor[row,column]
'''
```

```python
import torch
vector = torch.tensor([[10,20,30],[40,50,60],[70,80,90]])
print(vector.ndim)
print(vector.shape)
```

```python
print("element at (0,1)",vector[0])
print("Access first column",vector[:,0])
```

```python
# slicing operation
print("Elements from first two rows ",vector[:-1,:])
```

```python
print("First two columns:",vector[:,:-1])
```

```python
# Fancy Indexing
# (0,1)-->20
# (2,2)-->90

print("FancyIndexing:",vector[[0,2],[1,2]])
```

```python
# Boolean Indexing
mask = vector > 50
print(mask)
print(vector[mask])
```

```python
# Changing tensor values via indexing
print(vector)
vector[0,1] = 25
print(vector)
```

```python
vector[:,0] = torch.tensor([100,200,300])
print(vector)
```

```python
indices = torch.tensor([0,2])
torch.index_select(vector,dim=0,index=indices)
```

```python
# Access tensors with steps
print(vector[:,::2])
```

```python
torch.flip(vector,dims=(0,))
```
