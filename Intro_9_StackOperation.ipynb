STACK OPERATION

```python
import torch
```

```python
tensor_1 = torch.tensor([1,2,3])
tensor_2 = torch.tensor([4,5,6])
print(tensor_1.ndim)
print(tensor_1.size())
```

```python
stacked_tensor_dim0 = torch.stack((tensor_1,tensor_2),dim=0)
print(stacked_tensor_dim0.shape)
print(stacked_tensor_dim0)
```

```python
stacked_tensor_dim0 = torch.stack((tensor_1,tensor_2),dim=1)
print(stacked_tensor_dim0.shape)
print(stacked_tensor_dim0)
```

```python
tensor_3 = torch.tensor([7,8,9])
stacked_tensor = torch.stack((tensor_1,tensor_2,tensor_3),dim=1)
print(stacked_tensor.shape)
print(stacked_tensor)
```

```python
tensor_2d_1 = torch.tensor([[1,2],[3,4]])
tensor_2d_2 = torch.tensor([[5,6],[7,8]])
stacked_2d_tensor = torch.stack((tensor_2d_1,tensor_2d_2),dim=0)
print(stacked_2d_tensor.shape)
print(stacked_2d_tensor)
```

```python
stacked_2d_tensor2 = torch.stack((tensor_2d_1,tensor_2d_2),dim=1)
print(stacked_2d_tensor2.shape)
print(stacked_2d_tensor2)
```

```python
stacked_2d_tensor3 = torch.stack((tensor_2d_1,tensor_2d_2),dim=2)
print(stacked_2d_tensor3.shape)
print(stacked_2d_tensor3)
```

CONCATENATION OPERATION

```python
concatenated_tensors = torch.cat((tensor_2d_1,tensor_2d_2),dim=0)
print(concatenated_tensors.shape)
print(concatenated_tensors.ndim)
print(concatenated_tensors)
```

```python
concatenated_tensors = torch.cat((tensor_2d_1,tensor_2d_2),dim=1)
print(concatenated_tensors.shape)
print(concatenated_tensors.ndim)
print(concatenated_tensors)
```
