```python

import cv2
import easyocr
from ultralytics import YOLO

# 1. Initialize the models
# Note: 'yolov8n.pt' is a general model. For better results,
# use a weight file specifically trained on license plates.
detector = YOLO('/content/drive/MyDrive/YOLO_Training/number_plate_custom_model-7/weights/best.pt')
reader = easyocr.Reader(['en'], gpu=True) # Set gpu=True if you have a CUDA GPU
cap=cv2.VideoCapture("/content/drive/MyDrive/Off-Road.MP4")



# Get video properties for the writer
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = int(cap.get(cv2.CAP_PROP_FPS))
fourcc = cv2.VideoWriter_fourcc(*'mp4v') # Codec for .mp4

out = cv2.VideoWriter("/content/drive/MyDrive/ANPR/output.MP4", fourcc, fps, (width, height))




def process_license_plate(img):
    #img = cv2.imread(image_path)
    print("inside image processing function")
    # 2. Detect objects (looking for 'car', 'bus', 'truck', or custom plate class)
    # If using a custom plate model, classes=[0] usually refers to the plate.
    results = detector(img)[0]

    for result in results.boxes.data.tolist():
        x1, y1, x2, y2, score, class_id = result

        # Crop the detected bounding box
        plate_crop = img[int(y1):int(y2), int(x1):int(x2)]

        if plate_crop.size > 0:
            # 3. Perform OCR on the cropped plate
            ocr_result = reader.readtext(plate_crop)

            if ocr_result:
                # Extract the text with the highest confidence
                plate_text = ocr_result[0][-2].upper()

                # 4. Visualization
                # Draw Bounding Box
                cv2.rectangle(img, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)

                # Draw Text Background & Label
                cv2.putText(img, plate_text, (int(x1), int(y1) - 10),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.9, (36, 255, 12), 2)

    out.write(img)
    #cv2_imshow(img)
    #cv2.waitKey(0)
    #cv2.destroyAllWindows()

# Run the function
while cap.isOpened():
    print("captured stream opened")
    success,frame = cap.read()
    if success:
        print("Frame read successfully")
        process_license_plate(frame)
    else:
        print("Can't read input video")
        break
cap.release()
out.release()
```
